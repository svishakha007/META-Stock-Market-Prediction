Getting Historical Data for a Specific Period

```python
import datetime
import yfinance as yf
start_date = datetime.datetime(2019, 5, 31)
end_date = datetime.datetime(2021, 1, 30)
meta = yf.Ticker("META")
data = meta.history(start=start_date, end=end_date)
print(data.to_string())
```

```python
# Create a copy of the data
df = data.copy()

# Create technical indicators
df["MA_5"] = df["Close"].rolling(5).mean()
df["MA_20"] = df["Close"].rolling(20).mean()

# Daily return
df["Return"] = df["Close"].pct_change()

# Tomorrow's closing price
df["Tomorrow_Close"] = df["Close"].shift(-1)

# Target variable
# 1 = price goes up
# 0 = price goes down
df["Target"] = (df["Tomorrow_Close"] > df["Close"]).astype(int)

# Remove missing values
df = df.dropna()

print(df.head())
```

```python
#train the model

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Select input features
features = [
    "Open",
    "High",
    "Low",
    "Close",
    "Volume",
    "MA_5",
    "MA_20",
    "Return"
]

X = df[features]
y = df["Target"]

# 80% training, 20% testing
split = int(len(df) * 0.80)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]

# Create model
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train model
model.fit(X_train, y_train)

# Make predictions
predictions = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, predictions)

print("Model Accuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, predictions))
```

```python
# Get the latest available data
#predict the next market movement:
latest = X.iloc[[-1]]

prediction = model.predict(latest)[0]
probability = model.predict_proba(latest)[0]

if prediction == 1:
    print("\nPredicted Movement: UP")
else:
    print("\nPredicted Movement: DOWN")

print("Probability of DOWN:", round(probability[0] * 100, 2), "%")
print("Probability of UP:", round(probability[1] * 100, 2), "%")
```
